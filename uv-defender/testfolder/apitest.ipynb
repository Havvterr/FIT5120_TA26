```python
import requests

def test_weatherapi(api_key, city):
    url = "http://api.weatherapi.com/v1/current.json"  # Current weather request URL for WeatherAPI
    params = {
        'key': api_key,  # API key
        'q': city,       # Name of the city to query
        'aqi': 'no'      # No air quality information needed
    }

    response = requests.get(url, params=params)

    if response.status_code == 200:
        print("WeatherAPI test successful:")
        print(response.json())  # Print the returned JSON data
    else:
        print("WeatherAPI test failed:")
        print(f"Status Code: {response.status_code}, Response: {response.text}")

if __name__ == "__main__":
    # Replace with your API key and city name
    weather_api_key = "d47c344a5ef0415996334746250803"  # Your WeatherAPI key
    city_name = "London"                       # City name for testing
    
    # Test API
    test_weatherapi(weather_api_key, city_name)
```
