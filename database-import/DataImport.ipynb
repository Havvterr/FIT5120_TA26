## Load loacation csv data into as data frame


```python
import pandas as pd

# file path
file_path = 'australian_postcodes.csv'

# load csv and change datatype of postcode as str
df = pd.read_csv(file_path, dtype={'postcode': str})

# keep the following column
columns_to_keep = ['postcode', 'locality', 'state', 'long', 'lat']
df_filtered = df[columns_to_keep]

# rename the column
df_filtered.rename(columns={'long': 'longtitude', 'lat': 'latitude'}, inplace=True)

# make sure the datatype of postcode is str
df_filtered['postcode'] = df_filtered['postcode'].astype(str)

# make sure the longtitude and long
df_filtered['longtitude'] = df_filtered['longtitude'].astype(float)
df_filtered['latitude'] = df_filtered['latitude'].astype(float)

# print the head data 
print(df_filtered.head())

# check data type
print(df_filtered.info())

```
